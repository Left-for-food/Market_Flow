| **return_date** | datetime | Дата возврата заказа. Заполнена только для возвращенных заказов (`462` записи). Для остальных заказов содержит `NaN`. |
| **gross_margin_rub** | float64 | Оценка валовой прибыли с заказа в рублях. Для возвращенных заказов может принимать небольшие отрицательные значения. Среднее значение составляет около 274.5 рублей. |

  - **Gross Profit uplift** — изменение валовой прибыли;
  - **Return Rate** — доля возвратов.
- Наличие отрицательных значений `gross_margin_rub` для возвращенных заказов позволяет учитывать влияние возвратов на реальную прибыльность рекомендаций.

# **2. Инициализация БД**

The following schema was designed for the Market_Flow project.

*The diagram is available via the [dbdiagram.io](https://dbdiagram.io/d/Market_Flow-6a8d6333fd15a881e5effe6e)*



```python
import sqlalchemy
import pandas as pd
from sqlalchemy import text
import seaborn as sns

from db_conn import create_engine_custom
from db_conn import delete_tables

# Creating the database and creating an engine for it
DB_name = 'market_flow'
engine, _ = create_engine_custom(DB_name)

Create_DB_queries = open('Create_Market_Flow.sql').read().split(';')

delete_tables(engine) # Deleting previous tables if present

with engine.begin() as conn:
    for query in Create_DB_queries:
        if query.strip():
            conn.execute(text(query))
    print('Queries are successfully executed')
```

```python
with engine.begin() as conn:
    users.to_sql('users', conn, if_exists='append')
    exp_assignment.to_sql('experiment_assignment', conn, if_exists='append')
    products.to_sql('products', conn, if_exists='append')
    sessions.to_sql('sessions', conn, if_exists='append')
    impressions.to_sql('impressions', conn, if_exists='append')
    clicks.to_sql('clicks', conn, if_exists='append')
    orders.to_sql('orders', conn, if_exists='append')
    support_tickets.to_sql('support_tickets', conn, if_exists='append')

print('Data is uploaded successfully')
```
